## experiment 2


```python
import random
synonyms={
    "good":["great","escellent","nice"],
    "amazing":["great","fantastic"],
    "excellent":["wonderful","fantastic"],
    "love":["like","enjoy"],
    "terrible":["awful","horrible"],
    "hate":["dislike"],
    "boring":["dull","uninteresting"]
}

def synonym_replace(sentence):
  words=sentence.split()
  result=[]
  for word in words:
    clean=word.lower().strip(".,!?")
    if clean in synonyms:
      result.append(random.choice(synonyms[clean]))
    else:
      result.append(word)
  return " ".join(result)

text="movie was excellent."
print("original",text)
print("Augmented",synonym_replace(text))
```

```python
def random_deletion(sentence, probability=0.2):
 words = sentence.split()
 if len(words) <= 2:
  return sentence
 kept = [w for w in words if random.random() > probability]
 if not kept:
  kept = [random.choice(words)]
 return " ".join(kept)
text = "The movie was really very good."
print("Original:", text)
print("Augmented:", random_deletion(text, 0.2))
```

## Task 2



```python
synonym_dict = {
    "good": ["great", "excellent", "nice", "decent"],
    "bad": ["poor", "awful", "terrible"],
    "excellent": ["great", "fantastic", "superb"],
    "amazing": ["wonderful", "fantastic", "incredible"],
    "love": ["like", "enjoy", "adore"],
    "hate": ["dislike", "despise"],
    "boring": ["dull", "uninteresting", "tedious"],
    "terrible": ["awful", "horrible", "dreadful"],
    "happy": ["glad", "pleased", "delighted"],
    "cheap": ["affordable", "inexpensive"],
    "expensive": ["costly", "pricey"],
}

def synonym_replace(sentence):
    words = sentence.split()
    result = []
    for word in words:
        clean = word.lower().strip(".,!?")
        if clean in synonym_dict:
            result.append(random.choice(synonym_dict[clean]))
        else:
            result.append(word)
    return " ".join(result)

s = "The food was good and the service was fast."
print("Original :", s)
print("Augmented:", synonym_replace(s))
```

## Task 3

```python
def random_deletion(sentence, probability=0.2):
    words = sentence.split()
    if len(words) <= 2:
        return sentence
    kept = [w for w in words if random.random() > probability]
    if not kept:
        kept = [random.choice(words)]
    return " ".join(kept)

test_sentence = "The service at this restaurant was really very good today."
for p in [0.10, 0.20, 0.30]:
    print(f"p={p}:", random_deletion(test_sentence, p))
```

## Task 4


```python
import random
random.seed(42)

def random_insertion(sentence, word):
    words = sentence.split()
    position = random.randint(0, len(words))
    words.insert(position, word)
    return " ".join(words)

def random_swap(sentence):
    words = sentence.split()
    if len(words) < 2:
        return sentence
    i, j = random.sample(range(len(words)), 2)
    words[i], words[j] = words[j], words[i]
    return " ".join(words)

s = "The movie was good."
print("Original :", s)
print("Insertion:", random_insertion(s, "really"))
print("Swap     :", random_swap(s))

s2 = "I really enjoyed this movie."
print()
print("Original :", s2)
print("Insertion:", random_insertion(s2, "truly"))
print("Swap     :", random_swap(s2))
```

## Task 5


```python
methods = [
    lambda s: synonym_replace(s),
    lambda s: random_deletion(s, 0.3),
    lambda s: random_insertion(s, "really"),
    lambda s: random_swap(s),
]
def augment_sentence(sentence):
    return random.choice(methods)(sentence)

pool = df["text"].tolist()
for i in range(15):
    src = random.choice(pool)
    print(i+1, ":", src, "-->", augment_sentence(src))
```

## Task 6 - 8


```python
from sklearn.model_selection import train_test_split

# Imbalanced dataset: 12 positive, 4 negative
full_data = {
    "text": [
        "I love this product.", "The movie was excellent.", "This film was amazing.",
        "The service was good.", "I really enjoyed the food.", "This is a wonderful day.",
        "I love this movie.", "The food was good.", "This book was excellent.",
        "I had a good experience.", "The staff was really good.", "This app works good.",
        "I hate this movie.", "The movie was terrible.", "This film was boring.",
        "The service was bad.",
    ],
    "label": ["positive"]*12 + ["negative"]*4,
}
full_df = pd.DataFrame(full_data)

# Split FIRST, before any augmentation
train_df, test_df = train_test_split(
    full_df, test_size=0.25, random_state=42, stratify=full_df["label"]
)
print("Train (before augmentation):\n", train_df["label"].value_counts())

# Augment only the minority class within train_df
minority_label = train_df["label"].value_counts().idxmin()
majority_count = train_df["label"].value_counts().max()
minority_texts = train_df[train_df["label"] == minority_label]["text"].tolist()

aug_methods = [synonym_replace, lambda s: random_deletion(s, 0.15),
               lambda s: random_insertion(s, "really"), random_swap]

balanced_rows, current_count, attempt = [], len(minority_texts), 0
while current_count < majority_count:
    src = minority_texts[attempt % len(minority_texts)]
    method = aug_methods[attempt % len(aug_methods)]
    balanced_rows.append({"text": method(src), "label": minority_label})
    current_count += 1
    attempt += 1

final_train_df = pd.concat([train_df, pd.DataFrame(balanced_rows)], ignore_index=True)
print("\nTrain (after balancing):\n", final_train_df["label"].value_counts())
```

## Task 9 & 10

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

def train_and_eval(train_data, test_data, name):
    vectorizer = CountVectorizer()
    X_train = vectorizer.fit_transform(train_data["text"])
    X_test = vectorizer.transform(test_data["text"])
    clf = LogisticRegression().fit(X_train, train_data["label"])
    preds = clf.predict(X_test)

    print(f"\n[{name}] train size = {len(train_data)}")
    print("  Accuracy :", accuracy_score(test_data["label"], preds))
    print("  Precision:", precision_score(test_data["label"], preds, average="macro", zero_division=0))
    print("  Recall   :", recall_score(test_data["label"], preds, average="macro", zero_division=0))
    print("  F1-score :", f1_score(test_data["label"], preds, average="macro", zero_division=0))

train_and_eval(train_df, test_df, "Baseline (original train)")
train_and_eval(final_train_df, test_df, "Augmented (balanced train)")
```
