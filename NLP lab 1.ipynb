## Experiment 17

```python
text = '''
Hello!!! nhi krni NLP lab.
Contact: hassam@bashir.ex
Phone: 0300-7778881
Visit: https://hassam.capital
'''
email = re.findall(r"[\w\.-]+@[\w\.-]+\.\w+", text)
phone = re.findall(r"03\d{2}-\d{7}", text)
url = re.findall(r"https?://\S+", text)
clean_text = re.sub(r"[^\w\s]", " ", text)
clean_text = re.sub(r"\s+", " ", clean_text).strip()
print("Email:", email)
print("Phone:", phone)
print("URL:", url)
print("Clean Text:", clean_text)
```
