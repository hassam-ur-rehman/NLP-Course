## Exercise 1: Tokenize a paragraph (5+ sentences) and count sentences and word tokens

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")

paragraph = "Natural language processing is a fascinating field. It combines linguistics and computer science. Tokenization is one of the first steps in any NLP pipeline. Stemming and lemmatization help normalize words. Many real world applications rely on these techniques."

sentences = sent_tokenize(paragraph)
words = word_tokenize(paragraph)

print("Sentences:", sentences)
print("Number of Sentences:", len(sentences))
print("Word Tokens:", words)
print("Number of Word Tokens:", len(words))
```

## Exercise 3: Apply Porter, Snowball, and Lancaster stemmers to at least 20 words

```python
from nltk.stem import PorterStemmer, SnowballStemmer, LancasterStemmer

porter = PorterStemmer()
snowball = SnowballStemmer("english")
lancaster = LancasterStemmer()

words = ["play", "playing", "played", "plays", "player", "studies", "studying", "studied",
         "connect", "connected", "connection", "connecting", "relational", "generously",
         "happiness", "happier", "running", "runner", "flies", "fairly", "national", "organization"]

for word in words:
    print(word, "->", porter.stem(word), "|", snowball.stem(word), "|", lancaster.stem(word))
```

## Exercise 5: Regex-based stemmer with at least five suffix rules

```python
import re

def regex_stem(word):
    word = word.lower()
    word = re.sub(r"(ational|ization|fulness|iveness|ing|edly|ed|ies|es|s)$", "", word)
    return word

words = ["playing", "played", "books", "cars", "studies", "flies", "organization", "national", "happiness", "quickly"]

for word in words:
    print(word, "->", regex_stem(word))
```

## Exercise 6: Find five words where Porter, Snowball, and Lancaster produce different outputs

```python

porter = PorterStemmer()
snowball = SnowballStemmer("english")
lancaster = LancasterStemmer()

candidate_words = ["generously", "national", "relational", "organization", "university",
                    "happiness", "fairly", "running", "flies", "multiply"]

different_words = []
for word in candidate_words:
    p, s, l = porter.stem(word), snowball.stem(word), lancaster.stem(word)
    if len({p, s, l}) == 3:
        different_words.append((word, p, s, l))

for word, p, s, l in different_words[:5]:
    print(word, "-> Porter:", p, ", Snowball:", s, ", Lancaster:", l)

print()
```

## Exercise 7: Examples of over-stemming and under-stemming

```python

stemmer = PorterStemmer()

over_stemming_examples = [("university", "universe"), ("organization", "organ")]
under_stemming_examples = [("alumnus", "alumni"), ("run", "ran")]

print("Over-stemming examples (unrelated words reduced to the same stem):")
for w1, w2 in over_stemming_examples:
    print(w1, "->", stemmer.stem(w1), "|", w2, "->", stemmer.stem(w2))

print()
print("Under-stemming examples (related words not reduced to the same stem):")
for w1, w2 in under_stemming_examples:
    print(w1, "->", stemmer.stem(w1), "|", w2, "->", stemmer.stem(w2))
```

## Exercise 8: Apply lemmatization and compare it with Porter stemming

```python
from nltk.stem import WordNetLemmatizer, PorterStemmer

nltk.download("wordnet")
nltk.download("omw-1.4")

lemmatizer = WordNetLemmatizer()
stemmer = PorterStemmer()

words = ["studies", "studying", "played", "playing", "running", "ran", "flies", "goes", "having", "does"]

print(f"{'Word':<15}{'Porter Stem':<15}{'Lemma':<15}")
for word in words:
    print(f"{word:<15}{stemmer.stem(word):<15}{lemmatizer.lemmatize(word, pos='v'):<15}")
```

## Lab Challenge: Full pipeline — sentence tokenization, word tokenization, punctuation removal, Porter stemming

```python
import string
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.stem import PorterStemmer

nltk.download("punkt")
nltk.download("punkt_tab")

paragraph = input("Enter a paragraph: ")

stemmer = PorterStemmer()
sentences = sent_tokenize(paragraph)

print("Original Text:", paragraph)
print("Sentences:", sentences)

for sentence in sentences:
    tokens = word_tokenize(sentence)
    clean_tokens = [token for token in tokens if token not in string.punctuation]
    stems = [stemmer.stem(token) for token in clean_tokens]
    print()
    print("Sentence:", sentence)
    print("Tokens:", tokens)
    print("Cleaned Tokens:", clean_tokens)
    print("Stems:", stems)
```
